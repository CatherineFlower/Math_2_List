# Листок 2. Задача 2.1
Представить выражения в алгебраической, тригонометрической и показательной формах.

```python
import sympy as sp
import cmath, math

def forms(z_exact, name):
    z = sp.nsimplify(sp.simplify(sp.expand_complex(z_exact)))
    re, im = sp.re(z), sp.im(z)
    r = sp.simplify(sp.sqrt(re**2 + im**2))
    phi = sp.simplify(sp.arg(z))
    print(f'{name})')
    print('  алгебраическая:     ', sp.simplify(z))
    print('  модуль r =', r, '≈', float(r))
    print('  аргумент φ =', phi, '≈', float(phi), f'({math.degrees(float(phi)):.4f}°)')
    print('  тригонометрическая: ', f'{r}*(cos({phi}) + i*sin({phi}))')
    print('  показательная:      ', f'{r}*exp(i*({phi}))')
    return z
```

## а) $\dfrac{(5+i)(7-6i)}{3+i}$

```python
I = sp.I
za = (5+I)*(7-6*I)/(3+I)
za_v = forms(za, 'а')
```

## б) $\dfrac{(1+i)^5}{(1-i)^3}$

```python
zb = (1+I)**5/(1-I)**3
zb_v = forms(zb, 'б')
```

## в) $\dfrac{(-1+i\sqrt3)^{15}}{(1-i)^{20}}+\dfrac{(-1-i\sqrt3)^{15}}{(1+i)^{20}}$

```python
zc = (-1+I*sp.sqrt(3))**15/(1-I)**20 + (-1-I*sp.sqrt(3))**15/(1+I)**20
zc_v = forms(zc, 'в')
```

## Численная проверка (`cmath`)

```python
s3 = math.sqrt(3)
vals = {
 'а': (5+1j)*(7-6j)/(3+1j),
 'б': (1+1j)**5/(1-1j)**3,
 'в': (-1+1j*s3)**15/(1-1j)**20 + (-1-1j*s3)**15/(1+1j)**20,
}
for k,v in vals.items():
    r, phi = cmath.polar(v)
    print(k, v, '| r =', r, '| φ =', phi, '| r*e^{iφ} =', cmath.rect(r, phi))
```

## Ответ
| | алгебраическая | тригонометрическая | показательная |

| б) | $2$ | $2(\cos0+i\sin0)$ | $2e^{0\cdot i}$ |
| в) | $-64$ | $64(\cos\pi+i\sin\pi)$ | $64e^{i\pi}$ |
